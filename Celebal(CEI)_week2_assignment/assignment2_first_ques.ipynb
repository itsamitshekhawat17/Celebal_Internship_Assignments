# Step 4: Create SQL Database

The dataset is loaded into a SQL database to perform SQL-based analysis.

```python
conn = sqlite3.connect("superstore.db")
```

```python
df.to_sql(
    "superstore",
    conn,
    if_exists="replace",
    index=False
)
```

# Step 5: Explore SQL Table

The table is explored using SQL queries.

```python
query = """
SELECT *
FROM superstore
LIMIT 10
"""

pd.read_sql(query, conn)
```

### Insight

The first ten records confirm that the data has been loaded successfully into the SQL database.

# Step 6: Filtering Using WHERE Clause

Filtering helps extract specific records based on business requirements.

```python
query = """
SELECT *
FROM superstore
WHERE Region = 'West'
"""

pd.read_sql(query, conn)
```

The query retrieves all sales transactions belonging to the West region.

```python
query = """
SELECT *
FROM superstore
WHERE Category = 'Technology'
"""

pd.read_sql(query, conn)
```

This query identifies transactions related to Technology products.

```python
query = """
SELECT *
FROM superstore
WHERE Sales > 1000
"""

pd.read_sql(query, conn)
```

These transactions represent high-value sales.

```python
query = """
SELECT *
FROM superstore
WHERE "Order Date" LIKE '%2017'
"""

pd.read_sql(query, conn)
```

# Step 7: Aggregation Using GROUP BY

Aggregation summarizes large datasets into meaningful business metrics.

```python
query = """
SELECT
Region,
ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore
GROUP BY Region
ORDER BY Total_Sales DESC
"""

pd.read_sql(query, conn)
```

This analysis identifies the highest revenue generating regions.

```python
query = """
SELECT
Category,
SUM(Quantity) AS Total_Quantity
FROM superstore
GROUP BY Category
ORDER BY Total_Quantity DESC
"""

pd.read_sql(query, conn)
```

This shows which product category sold the highest quantity of items.


```python
query = """
SELECT
Category,
ROUND(AVG(Sales),2) AS Avg_Sales
FROM superstore
GROUP BY Category
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
Category,
ROUND(SUM(Profit),2) AS Total_Profit
FROM superstore
GROUP BY Category
ORDER BY Total_Profit DESC
"""

pd.read_sql(query, conn)
```

# Step 8: Sorting and Ranking Analysis

Sorting helps identify top-performing products and categories.

```python
query = """
SELECT
"Product Name",
ROUND(SUM(Sales),2) AS Revenue
FROM superstore
GROUP BY "Product Name"
ORDER BY Revenue DESC
LIMIT 10
"""

pd.read_sql(query, conn)
```

These are the highest revenue-generating products in the dataset.

```python
query = """
SELECT
"Customer Name",
ROUND(SUM(Sales),2) AS Revenue
FROM superstore
GROUP BY "Customer Name"
ORDER BY Revenue DESC
LIMIT 10
"""

pd.read_sql(query, conn)
```

A small group of customers contributes a significant share of total revenue.

```python
query = """
SELECT
substr("Order Date",-4) AS Year,
SUM(Sales) AS Revenue
FROM superstore
GROUP BY Year
ORDER BY Year
"""

pd.read_sql(query, conn)
```

This analysis helps understand yearly sales performance.
