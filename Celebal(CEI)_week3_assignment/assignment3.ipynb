# Step 6: Subqueries


Sales greater than the average sales


```python
query = """
SELECT * FROM orders
WHERE Sales >
(SELECT AVG(Sales)
 FROM orders
)
"""
pd.read_sql(query, conn)
```

This query identifies orders whose sales value is greater than the overall average sales amount.

HIGHEST ORDER PER CUSTOMER

```python
query = """
SELECT * FROM orders t1
WHERE Sales =
(SELECT MAX(Sales)
FROM orders
WHERE customer_id = t1.customer_id
)
"""
pd.read_sql(query, conn)
```

# Step 7: Common Table Expressions (CTEs)


TOTAL SALES PER CUSTOMER

```python
query = """
WITH customer_sales AS
(SELECT
    customer_id,
    ROUND(SUM(Sales),2) AS total_sales
    FROM orders
    GROUP BY customer_id
)
SELECT *
FROM customer_sales
ORDER BY total_sales DESC
"""
pd.read_sql(query, conn)
```

The CTE calculates total sales generated by each customer.

# Step 8: Window Functions



```python
query = """
SELECT customer_id, Sales,
ROW_NUMBER() OVER(ORDER BY Sales DESC)
AS row_num
FROM orders
"""
pd.read_sql(query, conn) 
```

```python
query = """
SELECT customer_id, Sales,
RANK() OVER( ORDER BY Sales DESC)
AS sales_rank
FROM orders
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT customer_id, Sales,
DENSE_RANK() OVER( ORDER BY Sales DESC)
AS sales_rank
FROM orders
"""
pd.read_sql(query, conn)
```

top 3 customers based on total sales

```python
query = """
WITH customer_sales AS
(
SELECT customer_id,ROUND(SUM(Sales),2) AS total_sales
FROM orders
GROUP BY customer_id
)
SELECT *
FROM
(
SELECT customer_id, total_sales,
RANK() OVER(ORDER BY total_sales DESC )
AS sales_rank
FROM customer_sales
)
WHERE sales_rank <= 3;"""
pd.read_sql(query,conn)
```

# Step 9: Combined Business Analysis
JOIN + CTE + WINDOW FUNCTION

```python
query = """
WITH customer_sales AS
( SELECT customer_id, ROUND(SUM(Sales),2) AS total_sales
FROM orders
GROUP BY customer_id
)
SELECT
c.customer_name,
cs.total_sales,
RANK() OVER(ORDER BY cs.total_sales DESC)
AS customer_rank
FROM customer_sales cs
JOIN customers c
ON cs.customer_id = c.customer_id
"""
pd.read_sql(query, conn)
```


This query ranks customers based on total sales contribution

### Query 1: Top 5 Customers Based on Total Sales

```python
query = """
WITH customer_sales AS
(SELECT customer_id, ROUND(SUM(Sales),2) AS total_sales
FROM orders
GROUP BY customer_id
)
SELECT
t1.customer_name,
t2.total_sales
FROM customer_sales t2
JOIN customers t1
ON t2.customer_id = t1.customer_id
ORDER BY t2.total_sales DESC
LIMIT 5
"""

pd.read_sql(query, conn)
```

### Query 2: Bottom 5 Customers Based on Total Sales

```python
query = """
WITH customer_sales AS
(
SELECT customer_id, ROUND(SUM(Sales),2) AS total_sales
FROM orders
GROUP BY customer_id
)
SELECT
t1.customer_name,
t2.total_sales
FROM customer_sales t2
JOIN customers t1
ON t2.customer_id = t1.customer_id
ORDER BY t2.total_sales ASC

LIMIT 5
"""

pd.read_sql(query, conn)
```

### Query 3: Customers Who Made Only One Order

```python
query = """
SELECT
t1.customer_name,
COUNT(t2.order_id) AS total_orders
FROM orders t2
JOIN customers t1
ON t2.customer_id = t1.customer_id
GROUP BY t2.customer_id
HAVING COUNT(t2.order_id) = 1
"""

pd.read_sql(query, conn)
```

### Query 4: Customers With Above-Average Sales

```python
query = """
WITH customer_sales AS
(
SELECT customer_id, ROUND(SUM(Sales),2) AS total_sales
FROM orders
GROUP BY customer_id
)
SELECT
t1.customer_name,
t2.total_sales
FROM customer_sales t2
JOIN customers t1
ON t2.customer_id = t1.customer_id
WHERE t2.total_sales >
(
SELECT AVG(total_sales)
FROM customer_sales
)
ORDER BY t2.total_sales DESC
"""

pd.read_sql(query, conn)
```

### Query 5: Highest Order Value Per Customer

```python
query = """
SELECT
t1.customer_name,
MAX(t2.Sales) AS highest_order_value
FROM orders t2
JOIN customers t1
ON t2.customer_id = t1.customer_id
GROUP BY t2.customer_id
ORDER BY highest_order_value DESC
"""
pd.read_sql(query, conn)
```
