# Step 1: Generate Realistic Datasets
Creates customers, products, orders, order_items CSVs with some intentional bad data.

```python
import random
from datetime import datetime, timedelta
import pandas as pd
from faker import Faker

fake = Faker()
random.seed(42)
Faker.seed(42)

N_CUSTOMERS = 200
N_PRODUCTS = 60
N_ORDERS = 800
```

## Customers table

```python
segments = ['Regular', 'Premium', 'New']
customers = []

for i in range(1, N_CUSTOMERS + 1):
    row = {
        'customer_id': i,
        'name': fake.name(),
        'email': fake.email(),
        'signup_date': fake.date_between(start_date='-2y', end_date='today'),
        'segment': random.choice(segments)
    }
    customers.append(row)


for i in random.sample(range(N_CUSTOMERS), 10):
    customers[i]['email'] = None

for i in random.sample(range(N_CUSTOMERS), 5):
    customers[i]['name'] = None

customers += [customers[i].copy() for i in random.sample(range(N_CUSTOMERS), 6)]

customers_df = pd.DataFrame(customers)
customers_df.head()
```

## Products table

```python
categories = ['Electronics', 'Clothing', 'Home', 'Books', 'Toys', 'Sports']
products = []

for i in range(1, N_PRODUCTS + 1):
    row = {
        'product_id': i,
        'product_name': fake.word().title() + ' ' + random.choice(['Pro', 'Max', 'Lite', 'Plus', '']),
        'category': random.choice(categories),
        'price': round(random.uniform(5, 500), 2)
    }
    products.append(row)


for i in random.sample(range(N_PRODUCTS), 4):
    products[i]['price'] = -products[i]['price']

for i in random.sample(range(N_PRODUCTS), 3):
    products[i]['category'] = None

products_df = pd.DataFrame(products)
products_df.head()
```

## Orders table

```python
statuses = ['completed', 'pending', 'cancelled', 'refunded']
orders = []

for i in range(1, N_ORDERS + 1):
    cust_id = random.randint(1, N_CUSTOMERS)
    order_date = fake.date_between(start_date='-2y', end_date='today')
    row = {
        'order_id': i,
        'customer_id': cust_id,
        'order_date': order_date,
        'status': random.choice(statuses)
    }
    orders.append(row)


for i in random.sample(range(N_ORDERS), 8):
    orders[i]['customer_id'] = 9999 + i


for i in random.sample(range(N_ORDERS), 5):
    orders[i]['order_date'] = datetime.today().date() + timedelta(days=random.randint(1, 60))


for i in random.sample(range(N_ORDERS), 6):
    orders[i]['status'] = None


orders += [orders[i].copy() for i in random.sample(range(N_ORDERS), 10)]

orders_df = pd.DataFrame(orders)
orders_df.head()
```

## Order items table

```python
order_items = []
item_id = 1

for o in orders:
    n_items = random.randint(1, 4)
    for _ in range(n_items):
        row = {
            'order_item_id': item_id,
            'order_id': o['order_id'],
            'product_id': random.randint(1, N_PRODUCTS),
            'quantity': random.randint(1, 5),
            'unit_price': round(random.uniform(5, 500), 2)
        }
        order_items.append(row)
        item_id += 1


for i in random.sample(range(len(order_items)), 10):
    order_items[i]['order_id'] = 88888 + i


for i in random.sample(range(len(order_items)), 5):
    order_items[i]['quantity'] = -order_items[i]['quantity']

for i in random.sample(range(len(order_items)), 5):
    order_items[i]['unit_price'] = None

order_items_df = pd.DataFrame(order_items)
order_items_df.head()
```

## Save raw CSVs

```python
customers_df.to_csv('../data/raw/customers.csv', index=False)
products_df.to_csv('../data/raw/products.csv', index=False)
orders_df.to_csv('../data/raw/orders.csv', index=False)
order_items_df.to_csv('../data/raw/order_items.csv', index=False)

print('customers:', len(customers_df))
print('products:', len(products_df))
print('orders:', len(orders_df))
print('order_items:', len(order_items_df))
```
